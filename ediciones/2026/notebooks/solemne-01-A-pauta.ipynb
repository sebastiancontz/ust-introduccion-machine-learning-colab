### Errores comunes

- Contar 13 columnas descartadas por olvidar que `monto_neto` y `prioridad_num` también están nombradas: el paso pide completar esas dos, no solo las nominales.
- Dejar `estado_cobranza` o `monto_gestion_cobranza` en el grupo de las que se perdieron por omisión: son las sospechosas de fuga y corresponde que queden fuera.
- Resolver el paso 3 diciendo solo que `dias_credito` «es numérica», sin nombrar la segunda afirmación que hace numerar ni reportar los dos anchos.

### Alternativas aceptables

- Tratar `dias_credito` como numérica y reportar 16 columnas, si la justificación declara que los tres plazos están igualmente espaciados y que el supuesto es sobre el modelo.
- Codificar `prioridad_cobranza` con one-hot en vez de numerarla, reportando 17 columnas, si la justificación sostiene que la distancia entre baja, media y alta no es comparable. Renuncia a la información del orden y eso hay que decirlo.
- Dejar `fecha_emision` y `fecha_vencimiento` en el grupo de las que corresponde dejar fuera, si el motivo es que una fecha no entra cruda a un modelo y hay que derivar de ella una variable —el mes, o el plazo en días— que en este ejercicio no se construye.
- Nombrar el descarte de las categorías poco frecuentes en vez de la agrupación en «otros», si se declara qué se pierde con cada opción.

```python
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import FunctionTransformer, OneHotEncoder

RUTA = (
    "https://raw.githubusercontent.com/sebastiancontz/"
    "ust-introduccion-machine-learning-colab/main/ediciones/2026/"
    "datasets/cobranza_servicios.csv"
)
facturas = pd.read_csv(RUTA)
for columna in ("fecha_emision", "fecha_vencimiento", "fecha_pago"):
    facturas[columna] = pd.to_datetime(facturas[columna], errors="coerce")

datos = facturas.drop_duplicates()
datos = datos.loc[datos["fecha_pago"] >= datos["fecha_emision"]].copy()
print("registros: %d de %d, con %d columnas" % (len(datos), len(facturas), datos.shape[1]))

ORDEN_PRIORIDAD = {"baja": 0, "media": 1, "alta": 2}
OBJETIVO = "dias_atraso"

# paso 1: el reparto en las tres rutas
NOMINALES = ["segmento", "region", "ejecutivo"]
ORDINAL = "prioridad_cobranza"
NUMERICA = "monto_neto"

datos["prioridad_num"] = datos[ORDINAL].map(ORDEN_PRIORIDAD)
print("prioridad sin numerar:", int(datos["prioridad_num"].isna().sum()))

preprocesador = ColumnTransformer([
    ("nom", OneHotEncoder(handle_unknown="ignore"), NOMINALES),
    ("num", FunctionTransformer(np.log1p, feature_names_out="one-to-one"), [NUMERICA]),
    ("ord", "passthrough", ["prioridad_num"]),
])
# el preprocesador recibe todas las predictoras candidatas: sin el objetivo y sin la
# version de texto de la ordinal, que ya quedo numerada
X = datos.drop(columns=[OBJETIVO, ORDINAL])
matriz = preprocesador.fit_transform(X)
categorias = sum(datos[columna].nunique() for columna in NOMINALES)
print("entran %d columnas y salen %d" % (X.shape[1], matriz.shape[1]))
print("la salida es %d del one-hot + 1 del logaritmo + 1 de la ordinal" % categorias)
for columna in NOMINALES:
    print("  %s: %d categorias" % (columna, datos[columna].nunique()))

# paso 2: lo que remainder="drop" descarto sin avisar
print("remainder por defecto:", preprocesador.remainder)
nombradas = NOMINALES + [NUMERICA, "prioridad_num"]
sin_nombrar = [columna for columna in X.columns if columna not in nombradas]
print("descartadas sin aviso: %d" % len(sin_nombrar))
print("  ", sin_nombrar)
print("monto_total contra monto_neto, correlacion: %.4f" % datos["monto_neto"].corr(datos["monto_total"]))

# paso 3: los dos tratamientos de dias_credito
plazos = sorted(int(valor) for valor in datos["dias_credito"].unique())
print("dias_credito toma %d valores: %s" % (len(plazos), plazos))
print("ancho si entra como numerica: %d" % (matriz.shape[1] + 1))
print("ancho si entra con one-hot:   %d" % (matriz.shape[1] + len(plazos)))

# paso 4: alta cardinalidad contra identificador
print("rut_cliente: %d valores distintos en %d registros" % (
    datos["rut_cliente"].nunique(), len(datos)))
print("ancho si rut_cliente entra con one-hot: %d" % (
    matriz.shape[1] + datos["rut_cliente"].nunique()))
print("id_factura: %d valores distintos en %d registros" % (
    datos["id_factura"].nunique(), len(datos)))
```

```python
import pandas as pd

RUTA = (
    "https://raw.githubusercontent.com/sebastiancontz/"
    "ust-introduccion-machine-learning-colab/main/ediciones/2026/"
    "datasets/cobranza_servicios.csv"
)
facturas = pd.read_csv(RUTA)
for columna in ("fecha_emision", "fecha_vencimiento", "fecha_pago"):
    facturas[columna] = pd.to_datetime(facturas[columna], errors="coerce")

# hallazgo 1: duplicados exactos
a_eliminar = int(facturas.duplicated().sum())
involucrados = int(facturas.duplicated(keep=False).sum())
print("duplicados exactos: %d registros involucrados, %d se eliminan" % (involucrados, a_eliminar))

# hallazgo 2: fecha de pago anterior a la de emision
imposibles = facturas["fecha_pago"] < facturas["fecha_emision"]
print("fecha de pago anterior a la emision:", int(imposibles.sum()))
print("  ids:", sorted(facturas.loc[imposibles, "id_factura"]))

# hallazgo 3: los dos extremos de monto_neto
print("monto_neto: media %.0f, mediana %.0f, maximo %d, minimo %d" % (
    facturas["monto_neto"].mean(), facturas["monto_neto"].median(),
    facturas["monto_neto"].max(), facturas["monto_neto"].min()))
print("montos bajo 10.000 (cargados en miles):", int((facturas["monto_neto"] < 10000).sum()))
print("montos sobre 100 millones:", int((facturas["monto_neto"] > 100_000_000).sum()))

preparado = facturas.drop_duplicates()
preparado = preparado.loc[preparado["fecha_pago"] >= preparado["fecha_emision"]].copy()
print("registros tras preparar: %d (de %d)" % (len(preparado), len(facturas)))
```
